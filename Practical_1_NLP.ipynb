```python
import nltk

# Download required data
nltk.download('punkt')
nltk.download('wordnet')

from nltk.tokenize import (
    wordpunct_tokenize,
    TreebankWordTokenizer,
    TweetTokenizer,
    MWETokenizer
)

# Sentence
sentence1 = "It's true, Ms. Martha Jones! #Truth"

# Whitespace
print("Whitespace tokenization =", sentence1.split())

# Punctuation
print("Punctuation-based tokenization =", wordpunct_tokenize(sentence1))

# MWE Tokenizer (Fixed version)
tokenizer = MWETokenizer()
tokenizer.add_mwe(('Martha', 'Jones'))

# Using split instead of word_tokenize (no error)
print("MWE tokenization =", tokenizer.tokenize(sentence1.split()))

# Tweet tokenizer
tweet = TweetTokenizer()
print("Tweet tokenization =", tweet.tokenize(sentence1))

# Treebank tokenizer
tree = TreebankWordTokenizer()
print("Treebank tokenization =", tree.tokenize(sentence1))
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

sentence1 = "It's true, Ms. Martha Jones! #Truth"

print("Tokenized words:", word_tokenize(sentence1))
print("Tokenized sentences:", sent_tokenize(sentence1))
```

```python
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

stemmer = PorterStemmer()

sentence2 = "I played the play playfully as the players were playing in the play"

# list of tokenized words
token = word_tokenize(sentence2)

# stems of each word
stem_words = [stemmer.stem(word) for word in token]

# print results
for e1, e2 in zip(token, stem_words):
    print(e1.ljust(13), '-->', '\t', e2)
```
